- Break , Pass, Continue

# Break
- when specific condtion met , it allows to terminate the loop 

```python
for i in "suhas":
    if i=='h':
        break
    print(i)
```

```python
a=[1,2,3,4,9,5,0.9]

for j in a:
    if j==9:
        break
    print(j)
```

```python
a=[1,2,3,4,9,5,0.9]

for j in a:
    if j==9:
        break
    print(j+1)
```

# continue

- when specific condition met, it allows to terimnate the loop and control tranfered to next statement 

```python
for i in "suhas":
    if i=='h':
        continue
    print(i)
```

```python
a=[1,2,3,4,9,5,0.9]

for j in a:
    if j==9:
        continue
    print(j)
```
