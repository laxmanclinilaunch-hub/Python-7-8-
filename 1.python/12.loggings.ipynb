```python
import logging
import os
```

```python
log_dir="log"

log_file_name="main.log"



    
os.makedirs(log_dir, exist_ok=True)

log_path=os.path.join(log_dir, log_file_name)
```

```python
logging.basicConfig(filename=log_path, level=logging.INFO, format='%(asctime)s - %(levelname)s - %(message)s')
```

```python
try:
    with open("non_existent_file.txt", "r") as file:
        print(file.read())
except Exception as e:
    logging.error("Exception occurred", exc_info=True)
```

```python
logging.debug("This is a debug message that will be logged to the file.")
logging.info("This is an info message that will be logged to the file.")
logging.warning("This is a warning message that will be logged to the file.")
logging.error("This is an error message that will be logged to the file.")

logging.critical("This is a critical message that will be logged to the file.")
```

```python
def add(a,b):
    result = a + b
    logging.info(f"Adding {a} and {b}, result: {result}")
    return result

def subtract(a,b):
    result = a - b
    logging.info(f"Subtracting {b} from {a}, result: {result}")
    return result


def multiply(a,b):
    result = a * b
    logging.info(f"Multiplying {a} and {b}, result: {result}")
    return result    

def divide(a,b):
    if b == 0:
        logging.error("Attempted to divide by zero.")
        raise ValueError("Denominator cannot be zero.")
    result = a / b
    logging.info(f"Dividing {a} by {b}, result: {result}")
    return result
```

```python
add(10, 5)

subtract(10, 5)

multiply(10, 5)

divide(10, 5)
```
