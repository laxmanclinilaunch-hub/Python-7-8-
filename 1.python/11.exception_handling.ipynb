- this type of errors we can easily handle try ,except

```python
def divide_num(x,y):
    if y==0:
        raise Exception("Division by zero is not allowed.")
    else:
        return x/y
   
```

```python
try:
    result=divide_num(10,0)
    print(result)
except Exception as e:
    print("Error:", e)
```

```python
def divide_num(x,y):
    try:
        return x/y
    except ZeroDivisionError:
        print("Error: Division by zero is not allowed.")
        return None
```

```python
divide_num(10,0)
```

```python
divide_num(10,2)
```
