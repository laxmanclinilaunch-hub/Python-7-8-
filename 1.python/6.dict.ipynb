- Dictionaries ?


- Mapping of one to one relationship between key , value pair is called "Dictionary"

- Mapping means collection of objects 

```python
marvel_dict={ "Name" :'Thor', 'weapon': "hammer", 'place': "asgard"   , "super_heros" :['iron_man','captian_america']   }
```

- keys must be unique 

```python
marvel_dict['Name']
```

- methods

- Keys()  - this methods returns the list of keys in the dict object

```python
marvel_dict.keys()
```

```python
list(marvel_dict.keys())
```

- values()

- this method returns the list of values in the dict object

```python
marvel_dict.values()
```

```python
list(marvel_dict.values())
```

- items()

this method returns the list of the keys & values

```python
marvel_dict.items()
```

```python
list(marvel_dict.items())
```

- get()

It takes an argument and returns None if the key is not found in the dict  otherwise (if it found , we will get related value)
