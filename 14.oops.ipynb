- class method :

   - for writting class method , we have to use decorator **@classmethod** and also we use **cls** parameter which refer to class not instance 

```python
class hotstar1:
    new_show="salar"


    def __init__(self,mail, password):
        self.mail=mail
        self.password=password
        self.service="free"
        self.amount=0


    def get_password(self):
        return self.password


    @classmethod
    def get_show(cls):
        return cls.new_show

    @classmethod
    def set_new_show(cls,show_name):
        cls.new_show=show_name 
```

```python
hotstar1.new_show     # accessing class variable using class method 
```

```python
user=hotstar1("laxman@gmail.com",123)

```

```python
user.get_show()
```

- Abstraction

- it is the process of hiding the internal implementation details and exposing only the essential functionality to the user

- generally for impplementation of this using ABC and @abstractmethod

```python
from abc import ABC, abstractmethod

#abstract class
class atm(ABC):


    @abstractmethod
    def withdraw(self):
        pass


#child class
class sbiatm(atm):

    def __init__(self,balance):
        self.balance=balance



    def withdraw(self):
        pin=int(input("enter pin :"))

        amount=int(input("enter the amount"))


        if pin==1234:
            if amount<=self.balance:
                self.balance-=amount
                print("transcation successful")
                print("please collect your cash") 
                print("available balance:", self.balance)

            else:
                print("insuffieent balance")
        else:
            print("invalidpin")
                   



# object
obj=sbiatm(100000)

obj.withdraw()
```

```python
from abc import ABC, abstractmethod



class hospital(ABC):

    @abstractmethod
    def treatment(self):
        pass


class cardio(hospital):
    def treatment(self):
        print("treating heart related diseases")


class neuro(hospital):
    def treatment(self):
        print("treament related brain & nervous diseases")



obj_1=cardio() 


obj_1.treatment()
```

```python
obj2=neuro()

obj2.treatment()
```
