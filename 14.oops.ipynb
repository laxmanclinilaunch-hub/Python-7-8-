- class method :

   - for writting class method , we have to use decorator **@classmethod** and also we use **cls** parameter which refer to class not instance 

```python
class hotstar1:
    new_show="salar"


    def __init__(self,mail, password):
        self.mail=mail
        self.password=password
        self.service="free"
        self.amount=0


    def get_password(self):
        return self.password


    @classmethod
    def get_show(cls):
        return cls.new_show

    @classmethod
    def set_new_show(cls,show_name):
        cls.new_show=show_name 
```
