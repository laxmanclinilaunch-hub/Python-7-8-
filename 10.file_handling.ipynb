# How file I/O done in  mostly programming languages

       - open file

       - Read/write data

       - close the file 

       - writing to file 




```python
f=open("files/sample.txt","w")

f.write("hello world")

f.close()

```

```python
f=open("files/sample1.txt","w")

f.write("hello world")

f.write("\n hi suhas")

f.write("\n how are you")

f.close()
```

```python
f=open("files/sample1.txt","a")

f.write("\n i am fine")
f.close()
```

```python
L=['hello   \nworld   \nhi suhas   \nhow are you   \ni am fine']


f=open("files/sample.txt","w")

f.writelines(L)

f.close()
```

- Reading files

```python
f=open("files/sample.txt","r")

s=f.read()

print(s)

f.close()
```

```python
f=open("files/sample1.txt","r")

s=f.read()

print(s)

f.close()
```

```python
f=open("files/sample.txt","r")

s=f.read(5)

print(s)

f.close()
```

```python
f=open("files/sample.txt","r")

s=f.read(10)

print(s)

f.close()
```
