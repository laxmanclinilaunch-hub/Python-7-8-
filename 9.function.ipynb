```python
def calculate_area(l,w):
    """
    calculate area of rectangle
    parameters:
    length (l)[float]: the length of rectangle
    width (w)[float]: the width f rectangle

    return:
    float :  the area of the rectangle 
    """
    return l*w

def calculate_perimeter(l,w):
    """
        calculate perimeter of rectangle
        parameters:
        length (l)[float]: the length of rectangle
        width (w)[float]: the width f rectangle
    
        return:
        float :  the perimeter of the rectangle 
        """
    return 2*(l+w)


def print_shape(l,w):
    """
    it prints area & perimeter of rectangle
    parameters:
            length (l)[float]: the length of rectangle
            width (w)[float]: the width f rectangle
    returns:  area & perimeter of rectangle
    """

    area=calculate_area(l,w)
    perimeter=calculate_perimeter(l,w)

    print(f"Rectangle with length {l} and width {w}:")

    print(f"Area {area}")

    print(f"Perimeter {perimeter}")
```

```python
print_shape(5,2)
```

```python
print_shape(7,2)
```

```python
print_shape(27,12)
```

```python
def cal_area(l,w):
    return l*w


cal_area(5,2)
```

```python
def calculate_Area(*args):
    product=1

    for i in args:
        product=product*i

    return product
```

```python
calculate_Area(2,4,7,5,3,4,2,2,1,9,56,3,5)
```

```python
def sum_all(*args):
    return sum(args)
```
